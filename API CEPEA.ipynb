# Import dados

```python
import agrobr
from datetime import datetime
import pandas
```

```python
# 1. Pega a data atual de hoje dinamicamente
hoje = datetime.today()

# 2. Calcula o ano de 10 anos atrás
ano_passado = hoje.year - 5
```

```python
data_inicio = f"{ano_passado}-01-01"
data_fim = hoje.strftime('%Y-%m-%d')
```

```python
lista_produtos = await agrobr.cepea.produtos()
print(lista_produtos)
```

```python
# 4. Executa a chamada com as datas dinâmicas
df = await agrobr.cepea.indicador(
    'arroz',
    inicio=data_inicio,
    fim=data_fim
)

# Exibe o resultado para checar o range do histórico
print(f"Buscando dados de {data_inicio} até {data_fim}")
display(df.head())
display(df.tail())
```

# Exploração

```python
resultados_unicos = df['praca'].unique()
print(resultados_unicos)
```
