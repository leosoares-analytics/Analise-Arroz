## Imports

```python
import pandas as pd
import duckdb
```

## CEPEA

### Caminho CEPEA

```python
caminho_arquivo = r"D:/Arquivos Pessoais/Dados/Case Arroz/Dados/CEPEA_20260613180726.xlsx"
```

### Conectar DF

```python
df = pd.read_excel(caminho_arquivo)
```

### Connect DuckDB
#### Precisa criar um arquivo dde banco de dados .db

```python
caminho_banco = r"D:/Arquivos Pessoais/Dados/Case Arroz/Banco_de_Dados/Valor_arroz.db"
con = duckdb.connect(caminho_banco)
```

### Salvar DF no DuckDB

```python
con.execute("CREATE OR REPLACE TABLE preco_arroz AS SELECT * FROM df")

print("Dados importados com sucesso para o DuckDB!")
```

### Fechar Conexão

```python
con.close()
```
